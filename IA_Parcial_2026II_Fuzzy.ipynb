```python
# Valores para evaluar el sistema
temp_ex = np.linspace(20, 100, 81)
carga_ex = np.linspace(0, 100, 101)

# Crear la malla
x, y = np.meshgrid(temp_ex, carga_ex)

# Matriz donde se almacenará la salida del sistema Sugeno
z_surface = np.zeros_like(x, dtype=float)

# Evaluar el sistema para cada combinación de temperatura y carga
for i in range(x.shape[0]):
    for j in range(x.shape[1]):
        z_surface[i, j] = inferencia_sugeno(
            x[i, j],
            y[i, j]
        )
```

```python
fig = go.Figure(
    data=[
        go.Surface(
            x=x,
            y=y,
            z=z_surface,

            contours={
                "z": {
                    "show": True,
                    "usecolormap": True,
                    "highlightcolor": "white",
                    "project_z": True
                }
            },

            colorbar=dict(
                title='Riesgo'
            )
        )
    ]
)

fig.update_layout(
    title='Superficie de respuesta del sistema Sugeno',

    scene=dict(
        xaxis_title='Temperatura (°C)',
        yaxis_title='Carga de trabajo (%)',
        zaxis_title='Riesgo',

        zaxis=dict(
            range=[0, 100]
        ),

        camera=dict(
            eye=dict(
                x=1.5,
                y=1.5,
                z=1.2
            )
        )
    ),

    width=950,
    height=700
)

fig.show()
```

```python
fig = go.Figure(
    data=go.Contour(
        x=temp_ex,
        y=carga_ex,
        z=z_surface,
        contours=dict(
            showlabels=True
        ),
        colorbar=dict(
            title='Riesgo'
        )
    )
)

fig.update_layout(
    title='Mapa de riesgo del sistema Sugeno',
    xaxis_title='Temperatura (°C)',
    yaxis_title='Carga de trabajo (%)',
    width=850,
    height=600
)

fig.show()
```
